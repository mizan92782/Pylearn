### Plots
* Line
* Bar
* Scatter
* Histogram


```python
from matplotlib import pyplot as plt 
#https://matplotlib.org/stable/api/pyplot_summary.html

x=[1,5,7,8,9]
y=[3,6,5,8,45]


#create a plot for x,y
plt.plot(x,y)


```

```python
# bar plot
plt.bar(x,y)
```

```python
# hitorgram
plt.hist(y)
```

```python
# Sctter
plt.scatter(x,y)
```

```python
# level and title

# title for plot
plt.title("x,y realation")

plt.xlabel("x level")
plt.ylabel("y lebel ")
```

### show multiple plot in a plot

```python

plt.plot(x,y)


plt.bar(x,y)
plt.show()
```

### Bar plot

```python
x=["Bangladesgh","India","Japan"]
y=[30,45,20]
 
'''
  1 argu: value
  2 argu : Labels = object
  autotpct = mathmatical structur
  startangle =starting angle
'''

plt.pie(y,labels=x, autopct='%1.1f%%', startangle=90)
```

### Area plot


```python

# to show a area  we need to y axis to show their differen
x = [1, 2, 3, 4, 5]
y1, y2 = [10, 20, 15, 25, 30], [5, 12, 10, 20, 25]


# for two y axis
plt.fill_between(x,y1,y2,color='green',alpha=0.4,label="area 1-2")

#set color  or marker for line
# marker for marker (x,y) point

plt.plot(x,y1,label='Line1',marker='*')
plt.plot(x,y2,label='line 2',marker='.')


```
