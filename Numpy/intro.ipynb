# Array Creation in Numpy

```python
import numpy as np


"""create empty array"""
arr=np.empty(2,dtype=int)
print(arr)
```

```python
"""  2d  array"""

arr=np.empty([2,2],dtype=int)
print(arr)
```

### Zeros method

```python
# zeros method:

arr=np.zeros(2)
print(arr)

# ndm
arr=np.zeros([10,10])
print(arr)
```

### Arrange method

```python

#arrange with value 0->n

print(" 0 -> n : ")
arr=np.arange(10)
print(arr,"\n")


# value: x->y
print("x -> y : ")
arr=np.arange(10,20)
print(arr,"\n")


# value: x ->y increasing inteval
print("x -> y incresing interval : ")
arr=np.arange(10,100,10)
print(arr,'\n')

print("Reshape data : ")
arr=np.arange(10).reshape(2,5)
print(arr,'\n')



# constract 3D array
print()
arr=np.arange(24).reshape(2,3,4)
print(arr)
```

### LineSpace method


```python
# divide xy interval in specific equal position
arr=np.linspace(1,4,6)
print("divide data : ",arr,"\n"*2)

# make arr vlaue sin
print("sin value : " ,np.sin(arr),"\n"*2)


```

### Flatten mathod
use for combine  nd array in 1d array

```python
arr=np.array([[1,2,3],[4,5,6]])



#row flatten
print("Raw flatten   :  ",arr.flatten())
print("Colum flatten :  ",arr.flatten('F'),'\n')
```

### ones method


```python
arr= np.ones((10,10))
print("all one : \n",arr,"\n"*2)


```

### Eye() :
* make diagonal array

```python
arr=np.eye(4,4, dtype=int,)
print(arr,"\n"*2)


arr=np.eye(4,4, dtype=int,k=-1)
print("remove first diagonal : \n",arr)
```

### Full(): 
creat a full array with a value


```python
arr=np.full([2,4],3)
print(arr)
```

### FromReconds()
use for make recond

```python
arr=np.core.records.fromrecords(
  [["Mizan ",21,400000],
    ["sizan",22,80000],
    ["karim ",201,4000]], names='Name,age,salary'
)


print(arr.Name)
print(arr.age)
print(arr.salary)
print(arr)
```
