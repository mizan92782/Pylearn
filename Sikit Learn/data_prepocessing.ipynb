# Data Preprocessing
! As we are dealing with lots of data and that data is in raw form, before inputting that data
to machine learning algorithms, we need to convert it into meaningful data. This process
is called preprocessing the data. Scikit-learn has package named preprocessing for this
purpose. The preprocessing package has the following techniques:



# Binarisation 
for boolean result depend on threshold vlaue

```python
import numpy as np
# import precessing pak from sklearn
from sklearn import preprocessing

Input_data = np.array(
 [

 [2.1, -1.9, 5.5],
 [-1.5, 2.4, 3.5],
 [0.5, -7.9, 5.6],
 [5.9, 2.3, -5.8]
 
 ])


 #binarized data ,threshol=.5
 #data above .5 will true

binarized_data=preprocessing.Binarizer(threshold=.5).transform(Input_data);

#output Binarized data
print(binarized_data)
```

# Standart deviation
Standard deviation is a measure of the amount of variation or dispersion in a set of values. It quantifies how much individual data points in a dataset differ from the mean (average) of the dataset. In other words, it provides a way to understand the spread or distribution of values in a dataset.


```python
import numpy as np
from sklearn import preprocessing



Input_data = np.array([
    [2.1, -1.9, 5.5],
    [-1.5, 2.4, 3.5],
    [0.5, -7.9, 5.6],
    [5.9, 2.3, -5.8]])


'''
from numpy useing mean and std
axis=0 : mean of colum
axis=1 : mean of row'''

def  mean_std(input_data):
    # mean
    print("colum means : ",input_data.mean(axis=0))
    print("row means : ",input_data.mean(axis=1)) 


    # Standard deviation
    print("colum deviation : ",input_data.std(axis=0))
    print("row deviation : ",input_data.std(axis=1))



# find mean and std
mean_std(Input_data)






```

# Remove mean and std to make data cluster
using  Scale function

```python
data_scaled=preprocessing.scale(Input_data)

"""find clustering data"""
mean_std(data_scaled)
```
