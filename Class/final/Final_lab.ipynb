### Introduction

The CIFAR-10 dataset is a benchmark dataset widely used in the field of computer vision for image classification tasks. It comprises 60,000 32x32 color images in 10 classes, with each class containing 6,000 images. The dataset is divided into a training set of 50,000 images and a testing set of 10,000 images. Each image is labeled with one of the following classes: airplane, automobile, bird, cat, deer, dog, frog, horse, ship, or truck. The dataset is commonly employed to evaluate the performance of machine learning algorithms and deep learning models for image classification.


### Dataset Description

Training Set: Contains 50,000 images, each labeled with one of the 10 classes. This set is utilized to train machine learning models.

Testing Set: Consists of 10,000 images, also labeled with one of the 10 classes. It serves as a benchmark to evaluate the performance of trained models on unseen data.

Image Size: Each image in the dataset is a 32x32 pixel RGB image, meaning it has three color channels (Red, Green, Blue).

Classes: The dataset is divided into 10 classes, with each class representing a distinct object category.

Label Distribution: The dataset is evenly distributed across the classes, with each class containing an equal number of images.



### Experimental Setup

To demonstrate the use of the CIFAR-10 dataset for image classification, we will develop a Convolutional Neural Network (CNN) using Python and TensorFlow/Keras libraries. The CNN will be trained on the training set and evaluated on the testing set. We will follow these steps:


Import necessary libraries.
Load and preprocess the CIFAR-10 dataset.
Define the CNN architecture.
Compile the model with appropriate loss function and optimizer.
Train the model on the training data.
Evaluate the model on the testing data.
Analyze the performance of the model.

# Code

Import Libraries

```python
import tensorflow as tf
from tensorflow.keras import layers, models, datasets
```

Function to resize images

```python
def resize_images(images, size):
    resized_images = tf.image.resize(images, size)
    return resized_images
```

Load and preprocess the CIFAR-10 dataset

```python

# Load and preprocess the CIFAR-10 dataset with resizing
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()

# Change the data format from (height, width, channels) to (batch, height, width, channels)
train_images = train_images.reshape(-1, 32, 32, 3)
test_images = test_images.reshape(-1, 32, 32, 3)

train_images = resize_images(train_images, (28, 28))
test_images = resize_images(test_images, (28, 28))

train_images, test_images = train_images / 255.0, test_images / 255.0
```

Define the CNN architecture

```python
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10)
])
```

 Compile the model


```python
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])
```

Train the model

```python
history = model.fit(train_images, train_labels, epochs=10, validation_data=(test_images, test_labels))

```

Evaluate the model

```python
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print('\nTest accuracy:', test_acc)
```
